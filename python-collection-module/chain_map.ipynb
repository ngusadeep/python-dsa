ChainMap is used to create a list of dictionaries. The collections.ChainMap data structure
combines several dictionaries into a single mapping. Whenever a key is searched in the chainmap,
it looks through all the dictionaries one by one, until the key is not found:

```python
#import module
from collections import ChainMap
```

```python
#class collections.ChainMap(dict1, dict2)
#create dictionaries
dict1 = {
    "Sept" : "Month September" ,
    "Nov" : "Month November" ,
    "Dec" : "Month December"
}

dict2 = {
    "PERN" : "PostgreSQL Express React NodeJs",
    "ML" : "Machine Learning" , 
    "CNN" : "Convolutional Neural Networks"
}
#add dictionaries in the chain
chain = ChainMap(dict1 , dict2)
```

```python
#print the values and keys
#dictinaries
print(dict1, '\n')
print(dict2, '\n')
#chain keys
print(list(chain.keys()), '\n')
#chain values
print(list(chain.values()), '\n')
#findind chain keys
print(chain['Nov'], '\n')
print(chain['PERN'])
```
