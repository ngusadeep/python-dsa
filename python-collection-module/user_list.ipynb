A UserList is a container that wraps list objects. It can be used to extend the functionality of the
list data structure. Consider the example code below, where adding/removing a new data element
is not allowed in the list data structure:

```python
from collections import UserList
class List(UserList):
    def push(self , key):
        raise RuntimeError('Canot push item!')
    def pop(self , key):
        raise RuntimeError('Can not remmove item')
l = List([23,45,67])

l.push(3683)
continue
l.pop([1])

```

In the above code, a customized push function in the MyList class is created to add the functionality
to not allow you to insert o remove an element into the list variable.
